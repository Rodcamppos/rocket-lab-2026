```python
files = {
    "movies_info": "movies_info_TMDB_IMDB.csv",
    "movies_financials": "movies_financials_IMDB_TMDB.csv",
    "movies_metrics": "movies_metrics_IMDB_TMDB.csv",
    "credits_and_tags": "credits_and_tags_IMDB_TMDB.csv",
    "movies_reviews": "movies_reviews.csv",
}

base_path = "/Volumes/workspace/cinedata/landing_inputs/"

dfs = {}
for name, filename in files.items():
    df = (spark.read
          .option("header", True)
          .option("inferSchema", True)
          .csv(base_path + filename))
    dfs[name] = df
    print(f"\n===== {name} ({filename}) =====")
    print(f"Linhas: {df.count()} | Colunas: {len(df.columns)}")
    df.printSchema()
```

```python
from pyspark.sql import functions as F

spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.bronze")

base_path = "/Volumes/workspace/cinedata/landing_inputs/"

file_table_map = {
    "movies_info_TMDB_IMDB.csv": "workspace.bronze.tb_movies_info",
    "movies_financials_IMDB_TMDB.csv": "workspace.bronze.tb_movies_financials",
    "movies_metrics_IMDB_TMDB.csv": "workspace.bronze.tb_movies_metrics",
    "credits_and_tags_IMDB_TMDB.csv": "workspace.bronze.tb_credits_and_tags",
    "movies_reviews.csv": "workspace.bronze.tb_movies_reviews",
}

for filename, table_name in file_table_map.items():
    df = (spark.read
          .option("header", True)
          .option("inferSchema", True)
          .csv(base_path + filename)
          .withColumn("ingestion_datetime", F.current_timestamp()))
    (df.write
       .format("delta")
       .mode("append")
       .saveAsTable(table_name))
    print(f"OK: {table_name} -> {df.count()} linhas")
```
